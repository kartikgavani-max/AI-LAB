```python
def get_neighbors(state):

    neighbors = []

    zero_pos = state.index(0)
    row = zero_pos // 3
    col = zero_pos % 3

    moves = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    for dr, dc, move in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_pos = new_row * 3 + new_col

            new_state = list(state)

            new_state[zero_pos], new_state[new_pos] = \
                new_state[new_pos], new_state[zero_pos]

            neighbors.append((tuple(new_state), move))

    return neighbors


def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


def reconstruct_path(parent, move_taken, goal):
    path = []

    current = goal

    while current is not None:
        path.append((current, move_taken[current]))
        current = parent[current]

    path.reverse()
    return path


def dfs(initial, goal):
    stack = [initial]

    visited = set()

    parent = {
        initial: None
    }

    move_taken = {
        initial: "START"
    }

    while stack:

        current = stack.pop()

        if current in visited:
            continue

        visited.add(current)

        if current == goal:
            return reconstruct_path(
                parent,
                move_taken,
                current
            )

        for next_state, move in get_neighbors(current):

            if next_state not in visited:
                if next_state not in parent:
                    parent[next_state] = current
                    move_taken[next_state] = move

                stack.append(next_state)

    return None


initial_state = (
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
)

goal_state = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)

solution = dfs(initial_state, goal_state)

if solution:
    print("DFS Solution:")
    print("Number of moves:", len(solution) - 1)
    print()

    for state, move in solution:
        print("Move:", move)
        print_state(state)
else:
    print("No solution found.")


print(" Time Complexity : d**d")
```
