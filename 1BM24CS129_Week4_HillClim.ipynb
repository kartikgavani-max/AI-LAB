```python
from itertools import combinations

def cost(state):
    conflicts = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):


            if state[i] == state[j]:
                conflicts += 1

            elif abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def get_neighbours(state):
    neighbours = []

    for i, j in combinations(range(len(state)), 2):
        new_state = state.copy()
        new_state[i], new_state[j] = new_state[j], new_state[i]

        neighbours.append((i, j, new_state, cost(new_state)))

    return neighbours

def hill_climbing(initial_state):
    current = initial_state.copy()

    print("Initial state:", current)
    print("Initial cost :", cost(current))
    print("-" * 50)

    while True:

        neighbours = get_neighbours(current)

        print("\nCurrent state:", current)
        print("Current cost :", cost(current))

        print("\nNeighbours:")

        for i, j, state, c in neighbours:
            print(f"  Swap (x{i}, x{j}) -> {state}, cost = {c}")

        best = min(neighbours, key=lambda x: (x[3], x[0], x[1]))

        i, j, best_state, best_cost = best

        if best_cost >= cost(current):
            print("\nNo better neighbour found.")
            print("Hill climbing terminated.")
            break

        print(f"\nChoose swap (x{i}, x{j})")
        print("New state:", best_state)
        print("New cost :", best_cost)

        current = best_state

        if best_cost == 0:
            print("\nGoal configuration reached!")
            break

    return current

initial_state = [3, 1, 2, 0]

solution = hill_climbing(initial_state)

print("\n" + "=" * 50)
print("Final solution:", solution)
print("Final cost    :", cost(solution))
```
