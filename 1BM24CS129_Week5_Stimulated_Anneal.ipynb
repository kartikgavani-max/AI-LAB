```python
import random
import math

def calculate_cost(board):
    conflicts = 0
    n = len(board)

    for c1 in range(n):
        for c2 in range(c1 + 1, n):

            if (board[c1] == board[c2] or
                abs(board[c1] - board[c2]) == abs(c1 - c2)):
                conflicts += 1

    return conflicts


def display_board(board):
    n = len(board)

    for row in range(n):
        for col in range(n):
            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()


def simulated_annealing():
    n = 8
    current = [random.randrange(n) for _ in range(n)]

    T = 10.0
    alpha = 0.99
    min_temperature = 0.001
    max_steps = 10000

    current_cost = calculate_cost(current)

    for step in range(max_steps):
        if current_cost == 0:
            break

        if T <= min_temperature:
            break


        next_state = current.copy()
        col = random.randrange(n)

        new_row = random.randrange(n - 1)
        if new_row >= next_state[col]:
            new_row += 1

        next_state[col] = new_row

        next_cost = calculate_cost(next_state)


        deltaE = next_cost - current_cost

        if deltaE < 0:
            current = next_state
            current_cost = next_cost

        else:
            probability = math.exp(-deltaE / T)

            if random.random() < probability:
                current = next_state
                current_cost = next_cost


        T *= alpha

    return current, current_cost, step + 1


solution, cost, steps = simulated_annealing()

print("Final board:", solution)
print("Final cost:", cost)
print("Iterations:", steps)

if cost == 0:
    print("\nSolution found!")
    display_board(solution)
else:
    print("\nNo solution found in this run.")
```
