```python
def get_neighbors(state):

    neighbors = []

    zero_pos = state.index(0)

    row = zero_pos // 3
    col = zero_pos % 3

    moves = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    for dr, dc, move in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_pos = new_row * 3 + new_col

            new_state = list(state)

            new_state[zero_pos], new_state[new_pos] = \
                new_state[new_pos], new_state[zero_pos]

            neighbors.append(
                (tuple(new_state), move)
            )

    return neighbors


def print_state(state):

    for i in range(0, 9, 3):
        print(state[i:i + 3])

    print()


def depth_limited_search(
        current,
        goal,
        depth,
        path,
        visited,
        moves):

    if current == goal:
        return True

    if depth == 0:
        return False

    visited.add(current)

    for next_state, move in get_neighbors(current):

        if next_state not in visited:

            path.append(next_state)
            moves.append(move)

            if depth_limited_search(
                    next_state,
                    goal,
                    depth - 1,
                    path,
                    visited,
                    moves):

                return True

            path.pop()
            moves.pop()

    visited.remove(current)

    return False


def ids(initial, goal):

    depth = 0

    while True:

        path = [initial]
        moves = ["START"]

        visited = set()

        found = depth_limited_search(
            initial,
            goal,
            depth,
            path,
            visited,
            moves
        )

        if found:
            return path, moves

        depth += 1

initial_state = (
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
)

goal_state = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)

path, moves = ids(
    initial_state,
    goal_state
)

print("IDS Solution:")
print("Number of moves:", len(path) - 1)
print()

for i in range(len(path)):

    print("Move:", moves[i])

    print_state(path[i])
```
